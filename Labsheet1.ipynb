# LAB SHEET - 1


```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
import numpy as np
image_path = "/content/cute-red-panda.jpg"
img = cv2.imread(image_path, -1)
```

### 1. Displaying the Image

```python
print('Original Image:')
cv2_imshow(img)
```

### 2. Converting Original Image to Grayscale and Binary Image

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print('Grayscale Image:')
cv2_imshow(gray_img)

gray_original_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary_original_img = cv2.threshold(gray_original_img, 127, 255, cv2.THRESH_BINARY)

print('Binary Image from Original Color Image:')
cv2_imshow(binary_original_img)
```

### 3. Computing the Complement (Negative) of the Image.

```python
negative_img = cv2.bitwise_not(img)

print('Negative (Complement) Image:')
cv2_imshow(negative_img)
```

### 4. Resize, rotate and crop the Image

```python
#Resize
height, width = img.shape[:2]
new_width = int(width * 0.5)
new_height = int(height * 0.5)
resized_img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_AREA)
print(f'Original Dimensions: {width}x{height}')
print(f'Resized Dimensions: {new_width}x{new_height}')
cv2_imshow(resized_img)

#Rotate
(h, w) = img.shape[:2]
center = (w // 2, h // 2)
angle = 180
scale = 1.0
M = cv2.getRotationMatrix2D(center, angle, scale)
rotated_img = cv2.warpAffine(img, M, (w, h))
print(f'Image rotated by {angle} degrees.')
cv2_imshow(rotated_img)

#Crop
height, width = img.shape[:2]
start_row, end_row = int(height * 0.20), int(height * 0.75)
start_col, end_col = int(width * 0.20), int(width * 0.80)
cropped_img = img[start_row:end_row, start_col:end_col]
print(f'Image cropped from ({start_col},{start_row}) to ({end_col},{end_row})')
cv2_imshow(cropped_img)
```

```python
output_filename = 'saved_image.jpg'

cv2.imwrite(output_filename, img)
print(f'Image successfully saved as {output_filename}')
```

### 5. Splitting and Merging Color Channels (RGB)

```python
b, g, r = cv2.split(img)
print('Red Channel:')
plt.imshow(r, cmap='Reds')
plt.axis('off')
plt.show()

print('Green Channel:')
plt.imshow(g, cmap='Greens')
plt.axis('off')
plt.show()

print('Blue Channel:')
plt.imshow(b, cmap='Blues')
plt.axis('off')
plt.show()
```

```python
merged_img = cv2.merge([b, g, r])

print('Merged Image (should look like the original):')
cv2_imshow(merged_img)
```
